```python
# filtering a column using condition
# ===========================
# condition df["age"] > 50
df[df["age"] > 50] # df[condition] return only matching rows

above_50 = df[df["age"] > 50]
above_50["first_name"] 

# or

df[df["age"] > 50]["first_name"] # return first name of who has aged above 50 as a series

df[df["age"] > 50][["first_name","job"]] # return a dataframe of first name and job of those
# who have age above 50
```

```python
# filtering using multiple conditions
# ====================================
# age above 50 and location = india

df[(df["age"] > 50) & (df["location"] == "india")]

# select column first_name,location those who have age above 40 and location = china

df[(df["age"] > 40) & (df["location"] == "china")][["first_name","location"]]

# select who has location in india or uk
df[(df["location"] == "india") | (df["location"] == "uk")]
```

```python
# slicing
# ===============
df[0:5] # row wise slicing
df[10:20] # selecting entire  rows from 10 to 19
df["age"] # selecting entire column
df[10:20][["first_name","job"]]
df.loc[10:20,["first_name","job"]] # is a label-based indexed used to select,filter and update
# specific rows and columns in a DataFrame
df[["age","job"]][10:20]
```

```python
df.loc[df["age"] > 50,["last_name","job"]]
# get the name and age of people having age above 40
df.loc[df["age"] > 40,["first_name","age"]]

# get name and job of people having age gte 40 and lte 60

df.loc[(df["age"] >= 40) & (df["age"] <= 60),["first_name","job"]]

# get the first_name,age,location of people having
# age > 50 and location = india

df.loc[(df["age"] > 50) & (df["location"] == "india"),["first_name","age","location"]]

# get f_name,job of people having age >= 60 and job id "musician"

df.loc[(df["age"] >= 60) & (df["job"] == "Musician"),["first_name","job"]]

# get the location and job of people aged above 30 or aged below 60

df.loc[(df["age"] > 30) | (df["age"] < 60),["location","job"]]

# get the f_name and location of people having job musician or therapist

df.loc[(df["job"] == "Musician") | (df["job"] == "Therapist"),["first_name","location"]]

# df.loc[rows,columns]

```

```python
# updating using loc
# ====================
df.loc[df["age"] > 60,"job"] = "Senior"

# change the location as india for people having age above 50
df.loc[df["age"] > 50,"location"] = "india"
df
```

.loc is label-based (using row and column names),
while .iloc is integer-position-based (using numbers from 0 to length-1)

```python
# iloc
# ========================
# iloc[row,column]
df.iloc[3,4]

# slicing
# ===============
df.iloc[0:5,4]

# select name from 6 rows
df.iloc[0:7,1]
```

```python
# get the first_name,location from 5 rows
df.iloc[0:6,[1,5]]
# select age from all rows
df.iloc[:,3]

# select first_name and location from all rows
df.iloc[:,[1,5]]
```

```python
# select last row
df.iloc[-5:]
```

```python
# return the first name,job,location from rows 1,10,14
df.iloc[[1,10,14],[1,4,5]]
```

```python
# get the first name of first 10 rows
df.iloc[:10,1]
# get the job and location of people aged above 45
df.loc[df["age"] > 45,["job","location"]]
# get the first name and age of last 10 employees
df.iloc[-10:,[1,3]]
# get the age of 5th row
df.iloc[4,3]
# get the employees from 100 to 110
df.iloc[100:111]
# rows from 100 to 120,name + age
df.iloc[100:121,[1,3]]
# get the first name of people aged above 60
df.loc[df["age"] > 60,"first_name"]
# get the first name,last name,age of people having age btw 30 and 50
df.loc[(df["age"] >= 30) & (df["age"] <= 50),["first_name","last_name","age"]]
# change the location as india for the people having age btw 50 and 60
df.loc[(df["age"] >= 50) & (df["age"] <= 60),"location"] = "india" 
df
```

```python
# group by
# ===============
# splits a dataframe into groups based on one or more columns
# applies aggregate function (lika a sum or average),
# and combines the result back into a summary dataframe
df.groupby("location")["age"].mean()
```

```python
# return the max age in each job
df.groupby("job")["age"].max()
```

```python
# sum(),count(),min(),max(),mean(),size()

# .count() --> return the non missing values from the df given
# .size() --> return the total number of rows from the column given
```

```python
# get the total count of people in each job

df.groupby("job").count() # return the non missing values from the df given
df.groupby("job").size() # return the total number of rows from the column given

df.groupby("job")["age"].agg(["mean","min","max","size"])
# .agg( return the multiple aggregate function result)

df.groupby(["job","location"]).size()
df.groupby("job")["age"].mean().sort_values(ascending = False)

# get the avg age in job + loaction in a descending order
df.groupby(["job","location"])["age"].mean().sort_values(ascending = False)

# find the average and maximun age for each job
df.groupby("job")["age"].agg(["mean","max"])
```

```python
# create a bar chart with total number of people in each location
result = df.groupby("location")["id"].size()
result = df.groupby("location")["id"].count()
result = df["location"].value_counts()
result
```

```python
# get a ber chart representing total number of people in each job
result = df["job"].value_counts().head(10)
result
```

```python
plt.bar(result.index,result.values)
plt.title("count of people in each job")
plt.xticks(rotation = 90)
plt.xlabel("jobs")
plt.ylabel("count")
plt.show()
```

```python
# get the maximum age in each location and represent in bar chart
result = df.groupby("location")["age"].max()
result
plt.bar(result.index,result.values)
plt.title("max age")
plt.xlabel("countries")
plt.ylabel("max age")
plt.show()
```
