```python
import pandas as pd
```

```python
data = pd.read_csv("customer5_windows.csv")
df = pd.DataFrame(data)
df
```

```python
df.iloc[0:10]
df.iloc[-10:,:]
df.sample(10)
df.shape
df.columns
df.dtypes
df.info()
df.describe()
df["prof"].unique()
df["loc"].unique()

df[["fname","lname","salary"]]
df[df["salary"] > 100000]
df[df["age"] > 50]
df[df["prof"] == "Pilot"]
df[df["loc"] == "india"]
df.loc[(df["salary"] >= 50000) & (df["salary"] <= 100000)]
df.loc[(df["age"] < 30) & (df["salary"] > 50000)]
df.loc[(df["loc"] == "india") | (df["loc"] == "uk")]
df.loc[(df["prof"].str.contains("Computer")) & (df["salary"] >  50000)]
df.loc[df["prof"] != "Pilot"]

df.iloc[:10,:5]
df.iloc[20:31]
df.iloc[:,[1,4,6]]
df.loc[df["salary"] > 100000,["fname","lname","salary"]]
df.loc[(df["loc"] == "india") & (df["age"] > 40)]
df.loc[df["salary"] < 30000,"salary"] += 5000
df.loc[(df["prof"] == "Pilot") & (df["loc"] == "india"),"loc"] = "India"

df.sort_values("salary")
df.sort_values("salary",ascending = False).head(10)
df.sort_values(["loc","salary"],ascending = [True,False])
df["salary"].mean()
df["salary"].max()
df["salary"].min()
df["age"].mean()
df["loc"].value_counts()

df.groupby("prof")["salary"].mean()
df.groupby("loc")["salary"].max()
df.groupby("prof")["salary"].min()
df.groupby("prof").size()
df.groupby("loc")["salary"].mean().sort_values(ascending = False)
df.groupby("prof")[["salary","age"]].agg("mean")
```
