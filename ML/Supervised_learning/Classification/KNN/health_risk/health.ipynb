```python
import pandas as pd
```

```python
data = pd.read_csv("Lifestyle_and_Health_Risk_Prediction_Synthetic_Dataset (1).csv")
df = pd.DataFrame(data)
df
```

```python
# columns,shape,describe,info
df.columns,df.shape,df.describe(),df.info()
```

```python
# missing values
df.isna().sum()
# fillna() used to fill the missing values
# df["age"].fillna(df["age"].mean())
# df["age"].fillna(df["age"].median())
# df["exersice"].fillna(df["exercise"].mode())
# to fill missing values with the mode(the most frequent value)
```

```python
# split the target and features
X = df.drop("health_risk",axis = 1)
y = df["health_risk"]
X,y
X.dtypes
```

```python
# Encodind >>> converting the string values in X into numerical values
# the column proffession has multiple values we are choosing one hot encoding

X = pd.get_dummies(X)
X
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.3,random_state = 42)
```

```python
# age = 25 - 75
# weight =40 -130
# height = 130 - 197

# scaling
# ====================
# Feature scaling in machine learning is the process of changing numerical columns 
# to use the same value range
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X_train) # mean + std deviation
scaler.transform(X_train)
# scaler.fit_transform(X_train)

scaler.transform(X_test)
#if we allow to fit the X-test it may occur the data leakage
#  
# fit(): learns and stores stats like mean and standerd deviation fro your dataset,
# but does not change the data
# transform(): uses the pre-calculated stats to scale or change the data
```

```python
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors = 5)
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score

print(accuracy_score(y_test,y_pred))
```
