```python
import pandas as pd
```

```python
data = pd.read_csv("Iris.csv")
df = pd.DataFrame(data)
df
```

```python
# missind values
df.isna().sum()
```

```python
# split features and target
X = df[["SepalLengthCm","SepalWidthCm","PetalLengthCm","PetalWidthCm"]]
y = df["Species"]
X,y
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.3,random_state =42)

# test_size = 0.2
# considering 20% for testing from X and y and remaining 80% for training the model
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
model = KNeighborsClassifier() # creating a model(object) for Kneighbor Classifier = 5
model.fit(X_train,y_train) # train model using X-train and y_train
y_pred = model.predict(X_test)
print(accuracy_score(y_test,y_pred))
```
