```python
dfe = pd.get_dummies(df[['Location','Fuel_Type','Transmission','Owner_Type','Brand']],drop_first=True,dtype=int)
dfe
```

```python
df_new = pd.concat([df,dfe],axis=1)
df_new
```

```python
df_new.drop(["Unnamed: 0","Name","Location","Fuel_Type","Transmission","Owner_Type","New_Price","Brand"],axis=1,inplace=True)
df_new
```

```python
df_new['Mileage'] = df_new['Mileage'].str.replace("km/kg","")
df_new['Mileage'] = df_new['Mileage'].str.replace("kmpl","")
df_new['Engine'] = df_new['Engine'].str.replace("CC","")
df_new['Power'] = df_new['Power'].str.replace("bhp","")

# replacing null with 0
df_new['Mileage'] = df_new['Mileage'].str.replace("null","0")
df_new['Engine'] = df_new['Engine'].str.replace("null","0")
df_new['Power'] = df_new['Power'].str.replace("null","0")
df_new
```

```python
df_new.dtypes
```

```python
df_new['Mileage'] = df_new['Mileage'].astype(float)
df_new['Engine'] = df_new['Engine'].astype(float)
df_new['Power'] = df_new['Power'].astype(float)
df_new.dtypes
```

```python
# converting 0 values to missing values
# object can be replace only with object..hence replace with 0 as object
# then converting to numerical value and then converting to missing value
# missing value is a numerical data type
```

```python
df_new.loc[df_new.Engine == 0,'Engine'] = np.nan
df_new.loc[df_new.Mileage == 0,'Mileage'] = np.nan
df_new.loc[df_new.Power == 0,'Power'] = np.nan
```

```python
df_new.isna().sum()
```

```python
df_new['Mileage'] = df_new['Mileage'].fillna(df_new['Mileage'].mean())
df_new['Engine'] = df_new['Engine'].fillna(df_new['Engine'].mean())
df_new['Power'] = df_new['Power'].fillna(df_new['Power'].mean())
df_new['Seats'] = df_new['Seats'].fillna(df_new['Seats'].mode()[0])
```

```python
df_new.isna().sum()
```

```python
df_new.duplicated().sum()
```

```python
x = df_new.drop(['Price'],axis = 1)
y = df_new['Price']
```
