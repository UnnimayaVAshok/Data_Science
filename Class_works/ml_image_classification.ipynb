```python
# core parts
# pixcel values - 0 to 255
input = []
output = []
categories = ['Dog','Cat']
datadir = '/content/drive/MyDrive/Data'
```

```python
for i in categories:
  print("Loading Categories " + i)
  path = os.path.join(datadir,i)
  for img in os.listdir(path):
    img_read = imread(os.path.join(path,img))
    img_res = resize(img_read,(150,150,3))
    input.append(img_res.flatten())
    output.append(categories.index(i))
  print("Loading Completed " + i)
```

```python
input = np.array(input)
output = np.array(output)
```

```python
len(input)
```

```python
len(output)
```

```python
df = pd.DataFrame(input)
df
```

```python
df['target'] = output
df
```

```python
x = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
x
y
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state=42)
```

```python
# model creation using SVM (Support Vector Machine) for image classification

from sklearn.svm import SVC

model = SVC()
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix

score = accuracy_score(y_test,y_pred)
print(score)
print(confusion_matrix(y_test,y_pred))
```
