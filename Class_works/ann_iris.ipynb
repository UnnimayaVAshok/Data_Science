```python
df.drop(["Id"],axis=1,inplace=True)
df
```

```python
# output is object hence need to encode it, mandatory step

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df["Species"] = le.fit_transform(df["Species"])
df
```

```python
x = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
import tensorflow as tf
from keras.models import Sequential
from keras.layers import Dense
```

```python
ann = Sequential()
ann.add(Dense(3,activation='relu'))
ann.add(Dense(10,activation='relu'))
ann.add(Dense(6,activation='relu'))
ann.add(Dense(3,activation='softmax'))
```

```python
ann.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
ann.fit(x_train,y_train,epochs=120,batch_size=20)
```

```python
ann.evaluate(x_test,y_test)
```

```python
y_pred = ann.predict(scaler.transform([[5.8,3.5,1.4,0.2]]))
y_pred
output = np.argmax(y_pred)
output
final_output = le.inverse_transform([output])
final_output
```
